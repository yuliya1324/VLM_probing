```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
```

```python
def plot_single(results_dir: str, ax: plt.Axes, label: str = None):
    results_path = Path(results_dir) / "probe_results.json"
    with open(results_path) as f:
        data = json.load(f)

    layers = sorted(data["per_layer"].keys(), key=int)
    accs = [data["per_layer"][l]["accuracy"] for l in layers]
    layer_nums = [int(l) for l in layers]

    label = label or Path(results_dir).name
    ax.plot(layer_nums, accs, marker="o", markersize=3, label=label)

    best_layer = layer_nums[accs.index(max(accs))]
    best_acc = max(accs)
    ax.annotate(f"best: L{best_layer} ({best_acc:.3f})",
                xy=(best_layer, best_acc),
                xytext=(best_layer + 1, best_acc - 0.03),
                fontsize=8, color="gray",
                arrowprops=dict(arrowstyle="->", color="gray", lw=0.8))

    return data
```

```python
results_dir = "/Data/iuliia.korotkova/VLM_probing/spatial/vila/"
title = "Probe Accuracy Across Layers for Spatial task"
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

data = plot_single(results_dir, ax)

n_classes = len(data["classes"])
chance = 1.0 / n_classes
ax.axhline(y=chance, color="red", linestyle="--", alpha=0.5,
            label=f"chance ({chance:.2f}, {n_classes} classes)")

ax.set_xlabel("Layer")
ax.set_ylabel("Validation Accuracy")
ax.set_title(title)
ax.legend()
ax.grid(alpha=0.3)
ax.set_ylim(0, 1.05)

plt.tight_layout()

plt.show()
```
